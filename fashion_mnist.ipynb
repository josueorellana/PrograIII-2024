```python
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt
import numpy as np
```

```python
fashion_mnist = tf.keras.datasets.fashion_mnist
(train_images, train_labels), (test_images, test_labels) = fashion_mnist.load_data()
```

```python
train_images = train_images / 255.0
test_images = test_images / 255.0
```

```python
label_names = [
    "Camiseta", "Pantalón", "Suéter", "Vestido", "Abrigo",
    "Sandalia", "Camisa", "Zapatilla", "Bolso", "Botines"
]
```

```python
new_labels = [
    "Top", "Pants", "Pullover", "Dress", "Coat",
    "Sandal", "Shirt", "Sneaker", "Bag", "Boot"
]
```

```python
plt.figure(figsize=(10,10))
for i in range(25):
    plt.subplot(5,5,i+1)
    plt.xticks([])
    plt.yticks([])
    plt.grid(False)
    plt.imshow(train_images[i], cmap=plt.cm.binary)
    plt.xlabel(new_labels[train_labels[i]])  # Convertir a texto solo al mostrar
plt.show()
```

```python
model = models.Sequential([
    layers.Flatten(input_shape=(28, 28)),   # Aplanar las imágenes 28x28
    layers.Dense(128, activation='relu'),  # Capa oculta con 128 neuronas y ReLU
    layers.Dense(10, activation='softmax') # Capa de salida para 10 clases
])
```

```python
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
```

```python
history = model.fit(train_images, train_labels, epochs=10, validation_split=0.2)
```
