Leer archivos con la libreria de pandas

```python
#importar la libreria de pandas.
import pandas as pd
```

```python
#leer el archivo de notas
ruta = "/content/notas.csv"
notas = pd.read_csv(ruta, sep=";", encoding="utf-8")

print(notas["nota"])
```

```python
#Obtener el listado de alumno aprobados
aprobados = notas[notas["nota"]>=6]
print(aprobados)
```

```python
#Obtener el listado de reprobados.
reprobados = notas[notas["nota"]<6]
print(reprobados)
```

```python
#Cuantos (numero) aprobaron la materia.
print("Numero de Aprobados", len(aprobados))
cantidad_aprobados = notas.loc[notas["nota"]>=6].agg({"count"})

print("Cantdad de aprobados", cantidad_aprobados)
```

```python
#Cuantos (Numero) reprobaron la materia
cantidad_reprobados = reprobados.agg({"count"})
print("Cantdad de reprobados: ", cantidad_reprobados)
```

```python
#agrupar por notas.
grupo = notas.groupby(["nota"]).agg({
  'nombre': lambda x: ', '.join(map(str, x))
})
print(grupo)
```

Entreando una IA de conversor de temperatura

```python
import tensorflow as tf
import pandas as pd
import numpy as np

temp = pd.read_csv("/content/conversor.csv", sep=";", encoding="utf-8")
print(temp, temp["f"])
```

```python
#datos de entrenamiento
celsius = temp["c"]
fahrenheit = temp["f"]

#print(fahrenheit)
```

```python
#modelo.
modelo = tf.keras.Sequential()
modelo.add(tf.keras.layers.Dense(units=1, input_shape=[1]))
```

```python
#compilar nuestro modelo.
modelo.compile(optimizer=tf.keras.optimizers.Adam(1.0), loss="mean_squared_error")

#entrenamiento
epocas = modelo.fit(celsius, fahrenheit, epochs=200, verbose=0)
```

```python
#probar nuestra IA
g = int(input("Celsius: "))
grados = np.array([g])
resp = modelo.predict(grados)
print("El valor en Farenheits es: ", round(resp[0][0],2) )
```

**Convertir de kelvin a farenger**

```python
import tensorflow as tf
import pandas as pd
import numpy as np

temp = pd.read_csv("/content/convertir.csv", sep=";", encoding="utf-8")
print(temp, temp["f"])
```

```python
# Datos de entrenamiento
kelvin = temp["k"]
fahrenheit = temp["f"]
```

```python
#Normalizacion de los datos
kelvin_norm = (kelvin - kelvin.min()) / (kelvin.max() - kelvin.min())
fahrenheit_norm = (fahrenheit - fahrenheit.min()) / (fahrenheit.max() - fahrenheit.min())
```

```python
# Modelo
modelo = tf.keras.Sequential()
modelo.add(tf.keras.layers.Dense(units=1, input_shape=[1]))
```

```python
# Compilar el modelo
modelo.compile(optimizer=tf.keras.optimizers.Adam(0.01), loss="mean_squared_error")

# Entrenamiento
epocas = modelo.fit(kelvin_norm, fahrenheit_norm, epochs=550, verbose=0)
```

```python
# Probar la IA
k = float(input("Kelvin: "))

k_norm = (k - kelvin.min()) / (kelvin.max() - kelvin.min())
temperatura = np.array([k_norm])

resp_norm = modelo.predict(temperatura)
resp = resp_norm[0][0] * (fahrenheit.max() - fahrenheit.min()) + fahrenheit.min()

print("El valor en Fahrenheit es:", round(resp, 2))
```
